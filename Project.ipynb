```python
reference_date = df['order_date'].max() + pd.Timedelta(days=1)
print("Reference Date :",reference_date)
```

```python
rfm = df.groupby('customer_id').agg(
    last_purchase_date = ('order_date', 'max'),
    frequency = ('order_id', 'count'),
    monetory = ('revenue', 'sum')
).reset_index()
```

```python
rfm
```

```python
rfm['recency'] = (reference_date - rfm['last_purchase_date']).dt.days
```

```python
rfm.head(5)
```

```python
rfm[['recency','frequency','monetory']].describe()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['recency'],
    5,
    labels=[5,4,3,2,1]
).astype(int)

rfm['F_Score'] = 1

rfm['M_Score'] = pd.qcut(
    rfm['monetory'],
    5,
    labels=[5,4,3,2,1]
).astype(int)
```

```python
rfm
```

```python
def assign_segment(row):
    R = row["R_Score"]
    M = row["M_Score"]

    if R >= 4 and M >= 4:
        return "High-Value Customer"
    elif R >= 4 and M <= 2:
        return "New Customer"
    elif R >= 4 and M == 3:
        return "Potential Loyalists"
    elif R == 3 and M >= 4:
        return "Promissing Customer"
    elif R <= 2 and M >= 4 :
        return "At-risk Customer"
    elif R <= 2 and M <= 2:
        return "Lost Customer"
    else:
        return "Average Customer"

rfm['customer_segment'] = rfm.apply(assign_segment, axis=1)
```

```python
rfm
```

```python
customer_segment_summary = rfm.groupby('customer_segment').agg(
    avg_recency = ('recency', 'mean'),
    avg_frequency = ('frequency', 'mean'),
    avg_monetory = ('monetory', 'mean'),
    customer_count = ('customer_id', 'count')
).reset_index().sort_values(by='customer_count', ascending=False)
```

```python
train_size = int(len(monthly_sales)* 0.8)
train = monthly_sales[:train_size]
test = monthly_sales[train_size:]
```

```python
print(train.dtype)
print(train.isnull().sum())
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

model = ExponentialSmoothing(
    train,
    trend= "add",
    seasonal= "add",
    seasonal_periods= 12
)

model_fit = model.fit()
```

```python
predictions = model_fit.forecast(len(test))
```

```python
from sklearn.metrics import mean_squared_error

rmse = np.sqrt(mean_squared_error(test, predictions))
print("RMSE :",rmse)
```

```python
mape = np.mean(np.abs((test - predictions) / test)) * 100
print("MAPE :",mape)
```

```python
plt.figure(figsize=(14,6))
plt.plot(train, label="Train")
plt.plot(test, label="Test")
plt.plot(predictions, label = "Predicted", linestyle = "--")

plt.title("Salesforecast Vs Actual")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.tight_layout()
plt.show()
```
